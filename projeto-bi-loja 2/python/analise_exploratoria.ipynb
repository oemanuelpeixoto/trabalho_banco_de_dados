# Análise exploratória — Loja de Eletrônicos
Base sintética de 500 vendas em 2025. Os resultados são ilustrativos.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
base = Path.cwd().parent / "dados"
vendas = pd.read_csv(base / "vendas_dashboard.csv", parse_dates=["data_id"])
vendas.head()
```

## Qualidade dos dados

```python
print("Dimensão:", vendas.shape)
print("Nulos por coluna:")
print(vendas.isna().sum())
print("IDs duplicados:", vendas.venda_id.duplicated().sum())
print(vendas[["quantidade","preco_unitario","faturamento"]].describe())
```

## Indicadores

```python
print("Faturamento: R$",round(vendas.faturamento.sum(),2))
print("Vendas:",vendas.venda_id.nunique())
print("Ticket médio: R$",round(vendas.faturamento.sum()/vendas.venda_id.nunique(),2))
```

## Categorias e tendência mensal

```python
vendas.groupby("categoria").faturamento.sum().sort_values().plot.barh(title="Faturamento por categoria")
plt.xlabel("R$")
plt.show()
vendas.groupby("ano_mes").faturamento.sum().plot.line(marker="o",title="Faturamento mensal")
plt.ylabel("R$")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Valores atípicos: critério IQR (não removidos automaticamente)

```python
q1,q3=vendas.faturamento.quantile([.25,.75])
iqr=q3-q1
outliers=vendas[vendas.faturamento > q3+1.5*iqr]
print("Vendas acima do limite IQR:",len(outliers))
print("Podem ser compras legítimas de produtos mais caros; não foram excluídas.")
```
